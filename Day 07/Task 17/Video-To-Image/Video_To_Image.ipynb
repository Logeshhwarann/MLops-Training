```python
import cv2

cap = cv2.VideoCapture("sample.mp4")

framerate = int(cap.get(cv2.CAP_PROP_FPS))

framecount = 0
count = 0

while True:
    success, frame = cap.read()

    if not success:
        break

    frame = cv2.resize(frame, (1280, 720))

    framecount += 1

    if framecount == framerate:

        cv2.imwrite(f"sample_frame_{count}.jpg", frame)

        print("Saved:", count)

        count += 1
        framecount = 0

cap.release()
cv2.destroyAllWindows()

print("Done")
```

```python
import zipfile
import os

folder_path = "."

zip_name = "extracted_frames.zip"

with zipfile.ZipFile(zip_name, 'w') as zipf:

    for file in os.listdir(folder_path):

        if file.endswith(".jpg"):
            zipf.write(file)

print("ZIP file created successfully:", zip_name)
```
