```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv('ai4i2020.csv')
```

```python

X = df[[
    'Air temperature [K]',
    'Rotational speed [rpm]',
    'Torque [Nm]',
    'Tool wear [min]'
]]
y = df['Process temperature [K]']
```

```python
model_multi = LinearRegression()


model_multi.fit(X, y)
```

```python
predictions_multi = model_multi.predict(X)
print(predictions_multi)
```

```python
mse_multi = mean_squared_error(y, predictions_multi)

print("Multivariable Regression MSE:", mse_multi)
```

```python
print("\nModel Coefficients:")
for feature, coef in zip(X.columns, model_multi.coef_):
    print(f"{feature}: {coef}")
print("\nIntercept:", model_multi.intercept_)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(8,6))
plt.scatter(y, predictions_multi)
plt.xlabel("Actual Temperature")
plt.ylabel("Predicted Temperature")
plt.title("Actual vs Predicted Temperature")
plt.plot(
    [y.min(), y.max()],
    [y.min(), y.max()]
)
plt.show()
```
